```python
import numpy as np 
v = np.array(v)
# cards_in_pack = v[-1]
v = v[:np.where(v == "Total")[0][0]]
v = np.reshape(v, (int(len(v)/4), 4))
v
```

```python
def convert_pull_rates(val):
    if "packs" in val:
        num = val.split()[2].replace(",", "")
        return 1/float(num)
    elif "cards per pack" or "card per pack" in val:
        return float(val.split()[0])
    elif "cards" in val:
        return float(val.split()[0])
```

```python
vfunc = np.vectorize(convert_pull_rates, otypes=[float])
rates = vfunc(v[:, [1,2,3]])
v[:, [1,2,3]] = rates
v
```
